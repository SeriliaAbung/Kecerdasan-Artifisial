K — CONSTRAINTS
Analisis harus mengikuti prinsip dasar metode Ground Penetrating Radar dan konsep geofisika yang benar. Jangan mengarang nilai kecepatan, kedalaman, jenis objek, atau kondisi lapangan yang tidak terdapat pada data.

Interpretasi reflektor dan objek harus dilakukan berdasarkan bentuk, amplitudo, kemenerusan, dan konteks lapangan. Jelaskan ketidakpastian interpretasi apabila terdapat lebih dari satu kemungkinan penyebab suatu refleksi. Bedakan antara data radargram, hasil pengolahan, dan interpretasi.

Jangan memberikan kesimpulan yang melebihi kemampuan data GPR. Jika kualitas data, atenuasi, atau resolusi membatasi interpretasi, jelaskan keterbatasannya. Gunakan istilah teknis geofisika yang tepat, tetapi tetap berikan penjelasan yang mudah dipahami oleh mahasiswa geofisika.

O — OUTPUT
