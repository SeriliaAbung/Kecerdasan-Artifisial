## Eksekusi 1: Electrical Resistivity Tomography (ERT)

**Data sintetis:** penampang resistivitas 2D hasil survei Dipole-Dipole dengan 41 elektroda dan spasi 2,5 m (panjang lintasan 100 m, kedalaman model 25 m). Tujuan survei adalah mengenali profil pelapukan, kedalaman batuan dasar, dan zona rekahan. Model berisi tanah penutup, batuan lapuk yang lembap atau lempungan, batuan dasar yang makin dalam ke arah kanan, satu zona rekahan sempit di sekitar jarak 60 m, dan satu tubuh resistif terisolasi di sekitar jarak 25 m. Data dibuat untuk demonstrasi, bukan hasil survei lapangan.

**Tujuan eksekusi:** menampilkan distribusi resistivitas lateral dan vertikal, lalu mengaitkan rentang nilainya dengan kemungkinan material dengan tetap menyebutkan ketidakpastiannya.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm
from scipy.ndimage import gaussian_filter

# ============================================================
# 1. PARAMETER SURVEI (SINTETIS)
# ============================================================
# Konfigurasi Dipole-Dipole, 41 elektroda, spasi 2,5 m -> panjang lintasan 100 m
n_elektroda, spasi = 41, 2.5
x = np.arange(n_elektroda) * spasi              # posisi elektroda (m)
z = np.linspace(0, 25, 101)                     # kedalaman model (m)
X, Z = np.meshgrid(x, z)

# ============================================================
# 2. MODEL RESISTIVITAS SINTETIS (hasil inversi 2D yang disimulasikan)
#    Target: profil pelapukan, kedalaman batuan dasar, dan zona rekahan
# ============================================================
z1 = 3.0 + 0.5*np.cos(X/12)                        # dasar tanah/pelapukan atas
z2 = 9.0 + 0.12*X                                   # dasar batuan lapuk (makin dalam ke kanan)

rho = np.where(Z < z1, 120.0,                       # tanah penutup
      np.where(Z < z2, 45.0,                        # batuan lapuk lempungan dan lembap
               900.0))                              # batuan dasar segar
rekahan = (X >= 58) & (X <= 63) & (Z >= 2)          # zona rekahan subvertikal terisi air
rho[rekahan] = 30.0
bongkah = ((X-25)/5)**2 + ((Z-6)/2.5)**2 <= 1       # tubuh resistif terisolasi (mis. bongkah / rongga)
rho[bongkah] = 700.0

rng = np.random.default_rng(14)
rho = np.exp(gaussian_filter(np.log(rho), sigma=1.4) + 0.04*rng.normal(size=rho.shape))

# ============================================================
# 3. RINGKASAN NILAI RESISTIVITAS PER ZONA
# ============================================================
zona = {
    "A. Tanah penutup"                   : (Z < z1-0.9) & ~((X>20)&(X<30)),
    "B. Batuan lapuk (lembap/lempungan)" : (Z > z1+1.0) & (Z < z2-1.2) & ~((X>20)&(X<30)) & ~((X>54)&(X<67)),
    "C. Batuan dasar"                    : (Z > z2+1.2) & ~((X>54)&(X<67)),
    "D. Zona rekahan (x = 58-63 m)"      : (X >= 59) & (X <= 62) & (Z > 4) & (Z < z2),
    "E. Tubuh resistif (x ~ 25 m)"       : ((X-25)**2/9 + (Z-6)**2/2.0 <= 1),
}
print("ERT - DATA SINTETIS (Dipole-Dipole, 41 elektroda, spasi 2,5 m, panjang 100 m)")
print(f"Rentang resistivitas model : {rho.min():.1f}-{rho.max():.0f} ohm.m\n")
for nama, m in zona.items():
    p5, p50, p95 = np.percentile(rho[m], [5, 50, 95])
    print(f"{nama:36s} median {p50:6.1f} ohm.m  (P5-P95: {p5:.0f}-{p95:.0f})")
for xp in (10, 40, 90):
    ix = np.argmin(abs(x-xp)); iz = np.argmax(rho[:, ix] > 300)
    print(f"Kedalaman puncak batuan dasar (>300 ohm.m) di x = {xp:3d} m : ~ {z[iz]:.1f} m")
print("Catatan: nilai dibuat untuk demonstrasi, bukan hasil inversi lapangan.")

# ============================================================
# 4. VISUALISASI
# ============================================================
fig, ax = plt.subplots(1, 2, figsize=(13, 4.8), gridspec_kw={"width_ratios": [3, 1]})
im = ax[0].imshow(rho, extent=[x.min(), x.max(), z.max(), z.min()], aspect="auto",
                  cmap="Spectral_r", norm=LogNorm(vmin=20, vmax=900))
fig.colorbar(im, ax=ax[0], label="Resistivitas (ohm.m)")
ax[0].set_xlabel("Jarak lintasan (m)"); ax[0].set_ylabel("Kedalaman (m)")
ax[0].set_title("Penampang resistivitas 2D sintetis (hasil inversi simulasi)")
for xp, c in ((10, "C0"), (90, "C1")):
    ax[1].semilogx(rho[:, np.argmin(abs(x-xp))], z, c, label=f"x = {xp} m")
ax[1].invert_yaxis(); ax[1].grid(True, which="both", alpha=0.3)
ax[1].set_xlabel("Resistivitas (ohm.m)"); ax[1].set_ylabel("Kedalaman (m)")
ax[1].set_title("Profil vertikal 1D"); ax[1].legend()
plt.tight_layout(); plt.show()
```

**Interpretasi hasil:** Metode ERT digunakan untuk memperoleh gambaran dua dimensi sebaran resistivitas bawah permukaan di sepanjang lintasan 100 m. Pengukuran memakai konfigurasi Dipole-Dipole dengan 41 elektroda berspasi 2,5 m. Konfigurasi ini peka terhadap perubahan lateral dan struktur vertikal seperti rekahan, sehingga sesuai dengan tujuan survei, tetapi sinyalnya melemah pada kedalaman besar. ## Eksekusi 2: Seismik Refleksi

**Data sintetis:** satu CMP gather dengan 41 trace (offset 0 sampai 1000 m, spasi 25 m) yang berisi tiga reflektor dengan wavelet 30 Hz dan noise acak, serta satu penampang stack zero-offset sepanjang sekitar 1,5 km yang memuat reflektor melengkung, reflektor miring, dan satu patahan normal. Data dibuat untuk demonstrasi, bukan hasil survei lapangan.

**Tujuan eksekusi:** memperlihatkan alur dasar pengolahan seismik refleksi: analisis kecepatan (semblance), koreksi NMO, konversi kecepatan RMS ke kecepatan interval dengan rumus Dix, dan interpretasi struktur pada penampang stack.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import find_peaks
from scipy.ndimage import uniform_filter1d

# ============================================================
# 1. DATA SINTETIS: CMP GATHER DENGAN 3 REFLEKTOR
# ============================================================
rng = np.random.default_rng(31)
dt, nt = 0.002, 751
t = np.arange(nt)*dt                                  # waktu tempuh dua arah (s)
h = np.arange(0, 1025, 25.0)                          # offset 0-1000 m (41 trace)
t0_true = np.array([0.40, 0.75, 1.10])                # waktu tempuh zero-offset (s)
vrms_true = np.array([1800.0, 2200.0, 2600.0])        # kecepatan RMS (m/s), hanya untuk membangkitkan data
amp = np.array([1.0, 0.8, 0.7])
f0 = 30.0

def ricker(tau):
    a = (np.pi*f0*tau)**2
    return (1 - 2*a)*np.exp(-a)

gather = np.zeros((nt, len(h)))
for t0, v, A in zip(t0_true, vrms_true, amp):
    th = np.sqrt(t0**2 + (h/v)**2)
    for k in range(len(h)):
        gather[:, k] += A/(1 + h[k]/1500)*ricker(t - th[k])
gather += 0.06*rng.normal(size=gather.shape)

# ============================================================
# 2. ANALISIS KECEPATAN (SEMBLANCE)
# ============================================================
v_scan = np.arange(1400, 3401, 20.0)
def nmo(data, vfun):
    out = np.zeros_like(data)
    for k, off in enumerate(h):
        tn = np.sqrt(t**2 + (off/vfun)**2)
        out[:, k] = np.interp(tn, t, data[:, k], left=0, right=0)
    return out

sem = np.zeros((nt, len(v_scan)))
for j, v in enumerate(v_scan):
    c = nmo(gather, np.full(nt, v))
    num = uniform_filter1d(c.sum(axis=1)**2, 15)
    den = uniform_filter1d((c**2).sum(axis=1), 15)*len(h)
    sem[:, j] = num/(den + 1e-9)

smax = sem.max(axis=1)
pk, _ = find_peaks(smax, height=0.35, distance=50)
pk = pk[(t[pk] > 0.2)]
t_pick = t[pk]; v_pick = v_scan[sem[pk].argmax(axis=1)]

# ============================================================
# 3. NMO + STACK, KONVERSI DIX DAN KEDALAMAN
# ============================================================
vfun = np.interp(t, t_pick, v_pick)
corr = nmo(gather, vfun)
stack_trace = corr.mean(axis=1)

v_int = np.zeros(len(t_pick)); thick = np.zeros(len(t_pick)); tprev, vprev = 0.0, 0.0
for i, (tt, vv) in enumerate(zip(t_pick, v_pick)):
    v_int[i] = np.sqrt((vv**2*tt - vprev**2*tprev)/(tt - tprev))
    thick[i] = v_int[i]*(tt - tprev)/2
    tprev, vprev = tt, vv
depth = np.cumsum(thick)

print("SEISMIK REFLEKSI - DATA SINTETIS (CMP gather, 41 trace, offset 0-1000 m, wavelet 30 Hz)")
print(f"Jumlah reflektor terdeteksi pada semblance : {len(t_pick)}")
for i in range(len(t_pick)):
    print(f"  Reflektor {i+1}: t0 = {t_pick[i]:.3f} s | Vrms = {v_pick[i]:.0f} m/s | "
          f"Vint (Dix) = {v_int[i]:.0f} m/s | tebal = {thick[i]:.0f} m | kedalaman dasar = {depth[i]:.0f} m")

# ============================================================
# 4. PENAMPANG STACK SINTETIS (ZERO-OFFSET) DENGAN PERLAPISAN MIRING DAN PATAHAN
# ============================================================
ntr, dxs = 120, 12.5
xs = np.arange(ntr)*dxs
i_ = np.arange(ntr)
R1 = 0.40 + 0.008*np.sin(2*np.pi*(i_-60)/60)
R2 = 0.75 + 0.0012*(i_-60) + np.where(i_ >= 75, 0.02, 0.0)
R3 = 1.10 + np.where(i_ >= 75, 0.05, 0.0)
sec = np.zeros((nt, ntr))
for R, A in ((R1, 1.0), (R2, 0.8), (R3, 0.7)):
    for k in range(ntr):
        a = A*(0.5 if 73 <= k <= 76 else 1.0)           # amplitudo melemah di dekat patahan
        sec[:, k] += a*ricker(t - R[k])
sec += 0.08*rng.normal(size=sec.shape)

def pick_h(nominal, cols):
    w = (t > nominal - 0.06) & (t < nominal + 0.08)
    return np.mean([t[w][np.argmax(sec[w, c])] for c in cols])
tR2_L, tR2_R = pick_h(0.75, range(50, 66)), pick_h(0.80, range(82, 98))
tR3_L, tR3_R = pick_h(1.10, range(50, 66)), pick_h(1.15, range(82, 98))
print(f"\nPenampang stack: patahan terindikasi di sekitar trace 75 (x ~ {75*dxs:.0f} m)")
print(f"Throw waktu R3 ~ {1000*(tR3_R - tR3_L):.0f} ms -> throw kedalaman ~ {v_int[2]*(tR3_R - tR3_L)/2:.0f} m")
print("Catatan: data dibuat untuk demonstrasi, bukan hasil survei lapangan.")

# ============================================================
# 5. VISUALISASI
# ============================================================
fig, ax = plt.subplots(2, 2, figsize=(13, 10))
ext = [h.min(), h.max(), 1.5, 0]
ax[0, 0].imshow(gather, aspect="auto", cmap="gray", extent=ext, vmin=-0.6, vmax=0.6)
ax[0, 0].set_title("CMP gather mentah"); ax[0, 0].set_xlabel("Offset (m)"); ax[0, 0].set_ylabel("Waktu (s)")
im = ax[0, 1].imshow(sem, aspect="auto", cmap="jet", extent=[v_scan.min(), v_scan.max(), 1.5, 0])
ax[0, 1].plot(v_pick, t_pick, "w-o", ms=5)
ax[0, 1].set_title("Spektrum kecepatan (semblance) dan picking"); ax[0, 1].set_xlabel("Vrms (m/s)")
fig.colorbar(im, ax=ax[0, 1], label="Semblance")
ax[1, 0].imshow(corr, aspect="auto", cmap="gray", extent=ext, vmin=-0.6, vmax=0.6)
ax[1, 0].set_title("CMP gather setelah koreksi NMO"); ax[1, 0].set_xlabel("Offset (m)"); ax[1, 0].set_ylabel("Waktu (s)")
ax[1, 1].imshow(sec, aspect="auto", cmap="gray", extent=[0, xs.max(), 1.5, 0], vmin=-0.8, vmax=0.8)
ax[1, 1].set_title("Penampang stack sintetis (zero-offset)")
ax[1, 1].set_xlabel("Jarak (m)"); ax[1, 1].set_ylabel("Waktu (s)")
for a in ax.ravel(): a.set_ylim(1.5, 0)
plt.tight_layout(); plt.show()
```

**Interpretasi hasil:** Metode seismik refleksi digunakan untuk memetakan batas lapisan dan struktur bawah permukaan dari gelombang yang dipantulkan oleh bidang kontras impedansi akustik. Data contoh berupa CMP gather dengan 41 trace pada offset 0 sampai 1000 m dan wavelet berfrekuensi sekitar 30 Hz. ## Eksekusi 3: Geomagnet

**Data sintetis:** peta anomali medan magnet total seluas 200 x 200 m dengan spasi titik 5 m. Medan utama diasumsikan sekitar 44.500 nT dengan inklinasi minus 25 derajat dan deklinasi 0 derajat (nilai contoh untuk demonstrasi, bukan nilai IGRF lokasi tertentu). Data berisi dua sumber terkubur bermagnetisasi induksi (satu dangkal dan kuat, satu lebih dalam), tren regional, dan noise sekitar 2 nT. Data dibuat untuk demonstrasi, bukan hasil survei lapangan.

**Tujuan eksekusi:** memperlihatkan pemisahan anomali residual dari tren regional, reduksi ke kutub (RTP), penentuan posisi sumber, dan estimasi kedalaman dengan pemodelan ke depan sederhana.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import least_squares

# ============================================================
# 1. PARAMETER MEDAN GEOMAGNET (nilai contoh untuk demonstrasi)
# ============================================================
F0, inc, dec = 44500.0, np.radians(-25.0), np.radians(0.0)   # nT, inklinasi, deklinasi
f_hat = np.array([np.cos(inc)*np.sin(dec), np.cos(inc)*np.cos(dec), -np.sin(inc)])   # (E, N, Up)

def dipol(X, Y, x0, y0, zs, k):
    """Anomali medan total (nT) dari sumber dipol terinduksi di kedalaman zs."""
    rx, ry, rz = X-x0, Y-y0, zs*np.ones_like(X)
    r = np.sqrt(rx**2 + ry**2 + rz**2)
    cphi = (f_hat[0]*rx + f_hat[1]*ry + f_hat[2]*rz)/r
    return k*(3*cphi**2 - 1)/r**3

# ============================================================
# 2. DATA SINTETIS: PETA 200 x 200 m, SPASI 5 m, DUA SUMBER TERKUBUR
# ============================================================
rng = np.random.default_rng(17)
x = np.arange(0, 201, 5.0); y = np.arange(0, 201, 5.0)
X, Y = np.meshgrid(x, y)
S1 = dict(x0=60.0,  y0=100.0, zs=7.0,  k=1.1e5)       # sumber dangkal (kuat)
S2 = dict(x0=150.0, y0=60.0,  zs=14.0, k=2.8e5)       # sumber lebih dalam
dT = dipol(X, Y, **S1) + dipol(X, Y, **S2)
tren = 0.25*(X-100) - 0.10*(Y-100) + 35                # tren regional
data = dT + tren + 2.0*rng.normal(size=X.shape)        # noise 2 nT
print("GEOMAGNET - DATA SINTETIS (peta 200 x 200 m, spasi 5 m, F ~ 44.500 nT, I = -25 derajat, D = 0 derajat)")
print(f"Rentang anomali medan total (data mentah): {data.min():.1f} sampai {data.max():.1f} nT")

# ============================================================
# 3. KOREKSI TREN REGIONAL (bidang orde-1)
# ============================================================
A = np.c_[np.ones(X.size), X.ravel(), Y.ravel()]
coef, *_ = np.linalg.lstsq(A, data.ravel(), rcond=None)
resid = data - (A@coef).reshape(X.shape)
print(f"Rentang anomali residual                 : {resid.min():.1f} sampai {resid.max():.1f} nT")

# ============================================================
# 4. REDUCTION TO THE POLE (RTP) DI DOMAIN FREKUENSI
# ============================================================
ny, nx = resid.shape
kx = 2*np.pi*np.fft.fftfreq(nx, 5.0); ky = 2*np.pi*np.fft.fftfreq(ny, 5.0)
KX, KY = np.meshgrid(kx, ky)
theta = np.arctan2(KX, KY)
kmag = np.sqrt(KX**2 + KY**2)
den = (np.sin(inc) + 1j*np.cos(inc)*np.cos(dec - theta))**2
Hrtp = 1/den
Hrtp[0, 0] = 1
taper = np.exp(-(kmag/(0.55*kmag.max()))**4)               # low-pass agar tidak menguatkan noise
rtp = np.real(np.fft.ifft2(np.fft.fft2(resid)*Hrtp*taper))

# ============================================================
# 5. LOKASI SUMBER DAN ESTIMASI KEDALAMAN (fitting profil melintasi puncak RTP)
# ============================================================
def puncak_sumber(m, jarak_min=0):
    iy, ix = np.unravel_index(np.argmax(m), m.shape)
    return x[ix], y[iy], m[iy, ix]
xp1, yp1, a1 = puncak_sumber(rtp)
m2 = rtp.copy(); m2[(abs(X-xp1) < 40) & (abs(Y-yp1) < 40)] = -1e9
xp2, yp2, a2 = puncak_sumber(m2)
print(f"\nPuncak RTP 1: x = {xp1:.0f} m, y = {yp1:.0f} m, amplitudo {a1:.1f} nT")
print(f"Puncak RTP 2: x = {xp2:.0f} m, y = {yp2:.0f} m, amplitudo {a2:.1f} nT")

def fit_profil(xp, yp, zs0=10.0, jendela=50):
    iy = np.argmin(abs(y - yp)); xs = x[abs(x - xp) <= jendela]; ob = resid[iy, abs(x - xp) <= jendela]
    f = lambda p: dipol(xs, np.full_like(xs, yp), p[0], yp, p[1], p[2]) + p[3] - ob
    s = least_squares(f, [xp, zs0, 1e5, 0.0], bounds=([xp-20, 2, 1e3, -50], [xp+20, 40, 1e7, 50]))
    return xs, ob, s.x, np.sqrt(np.mean(s.fun**2))
res = []
for (xp, yp) in ((xp1, yp1), (xp2, yp2)):
    xs, ob, p, rms = fit_profil(xp, yp)
    res.append((xs, ob, p, yp))
    print(f"Fitting profil y = {yp:.0f} m: sumber di x ~ {p[0]:.1f} m, kedalaman ~ {p[1]:.1f} m, RMS misfit {rms:.1f} nT")
print("Catatan: parameter medan dan sumber dibuat untuk demonstrasi, bukan data lapangan.")

# ============================================================
# 6. VISUALISASI
# ============================================================
fig, ax = plt.subplots(2, 2, figsize=(12, 10))
lim = np.percentile(abs(resid), 99)
for a, m, jd in ((ax[0, 0], resid, "Anomali medan total residual (nT)"), (ax[0, 1], rtp, "Setelah reduksi ke kutub / RTP (nT)")):
    im = a.pcolormesh(X, Y, m, cmap="RdBu_r", vmin=-lim, vmax=lim, shading="auto")
    fig.colorbar(im, ax=a, label="nT"); a.set_title(jd); a.set_aspect("equal")
    a.set_xlabel("X (m)"); a.set_ylabel("Y (m)")
for i, (xs, ob, p, yp) in enumerate(res):
    a = ax[1, i]
    xf = np.linspace(xs.min(), xs.max(), 200)
    a.plot(xs, ob, "ko", ms=4, label="Data residual")
    a.plot(xf, dipol(xf, np.full_like(xf, yp), p[0], yp, p[1], p[2]) + p[3], "r-", label=f"Model dipol (z ~ {p[1]:.1f} m)")
    a.grid(True, alpha=0.3); a.set_xlabel("X (m)"); a.set_ylabel("Anomali (nT)")
    a.set_title(f"Profil melintasi sumber {i+1} (y = {yp:.0f} m)"); a.legend()
plt.tight_layout(); plt.show()
```

**Interpretasi hasil:** Metode geomagnet digunakan untuk mengenali sumber magnetik terkubur dari anomali medan magnet total yang diukur di permukaan. Data contoh berupa peta seluas 200 x 200 m dengan spasi titik 5 m. Medan utama diasumsikan sekitar 44.500 nT dengan inklinasi minus 25 derajat, sehingga medan tidak vertikal dan anomali berbentuk dipol dengan pasangan lobus positif dan negatif. Data mentah berkisar dari sekitar minus 133 sampai 244 nT, dan sebagian variasinya berasal dari tren regional yang landai. Setelah tren dihilangkan dengan bidang orde pertama, anomali residual berkisar dari sekitar minus 158 sampai 219 nT.

Peta residual memperlihatkan dua kelompok anomali dipol. Di sekitar koordinat 60 m pada sumbu X dan 100 m pada sumbu Y terdapat anomali kuat dengan lobus positif di sisi utara sekitar 219 nT dan lobus negatif di sisi selatan sekitar minus 158 nT. Pasangan lobus yang rapat menunjukkan sumber yang dangkal. ## Eksekusi 4: Induced Polarization (IP)

**Data sintetis:** penampang resistivitas dan chargeability hasil survei Pole-Dipole sepanjang 150 m dengan kedalaman model 40 m. Model berisi tubuh mineralisasi sulfida tersebar (chargeability tinggi, resistivitas rendah) di bagian kiri-tengah, lapisan lempung dangkal di bagian kanan (resistivitas sangat rendah, chargeability rendah hingga sedang), dan zona batuan segar atau silisifikasi di bagian kiri (resistivitas tinggi, chargeability rendah). Data dibuat untuk demonstrasi, bukan hasil survei lapangan.

**Tujuan eksekusi:** menunjukkan bahwa anomali IP harus dibaca bersama resistivitas, karena kombinasi kedua parameter membedakan zona yang pada salah satu penampang tampak sama-sama anomali.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm, ListedColormap
from matplotlib.patches import Patch

# ============================================================
# 1. MODEL SINTETIS RESISTIVITAS & CHARGEABILITY (hasil inversi simulasi)
#    Pole-dipole, lintasan 150 m, kedalaman model 40 m
# ============================================================
x = np.linspace(0, 150, 151)
z = np.linspace(0, 40, 81)
X, Z = np.meshgrid(x, z)
g = lambda xc, zc, sx, sz: np.exp(-(((X-xc)/sx)**2 + ((Z-zc)/sz)**2)/2)

rng = np.random.default_rng(23)
rho = 100 + 1.5*Z                                               # latar belakang (ohm.m)
M   = 2.5 + 0.02*Z                                             # latar belakang (mV/V)

b_sulfida = g(55, 14, 12, 6)      # tubuh mineralisasi sulfida tersebar : resistivitas rendah, chargeability tinggi
b_lempung = g(118, 5, 20, 3.5)        # lapisan lempung dangkal            : resistivitas sangat rendah, chargeability rendah-sedang
b_silika  = g(20, 22, 14, 8)      # zona silisifikasi / batuan segar   : resistivitas sangat tinggi, chargeability rendah

rho = rho*(1 - 0.55*b_sulfida)*(1 - 0.85*b_lempung)*(1 + 4.0*b_silika)
M   = M + 22*b_sulfida + 3.5*b_lempung
rho *= np.exp(0.03*rng.normal(size=rho.shape))
M   *= (1 + 0.04*rng.normal(size=M.shape))

# ============================================================
# 2. KLASIFIKASI ZONA BERDASARKAN KOMBINASI RESISTIVITAS & CHARGEABILITY
# ============================================================
rho_rendah, rho_tinggi, M_tinggi = 50, 250, 10
kelas = np.zeros(rho.shape, dtype=int)                         # 0 = latar belakang
kelas[(M >= M_tinggi) & (rho < rho_tinggi)] = 1                # M tinggi, rho rendah-sedang -> sulfida tersebar
kelas[(M < M_tinggi) & (rho < rho_rendah)]  = 2                # M rendah, rho rendah       -> lempung / air asin
kelas[(M < M_tinggi) & (rho >= rho_tinggi)] = 3                # M rendah, rho tinggi       -> batuan segar / silika

print("IP - DATA SINTETIS (pole-dipole, lintasan 150 m, kedalaman model 40 m)")
print(f"Resistivitas  : {rho.min():.1f}-{rho.max():.0f} ohm.m (latar belakang sekitar 100-160 ohm.m)")
print(f"Chargeability : {M.min():.1f}-{M.max():.1f} mV/V (latar belakang sekitar 2,5-3,5 mV/V)\n")
nama = {1: "Anomali IP tinggi + resistivitas rendah", 2: "Resistivitas rendah + IP rendah",
        3: "Resistivitas tinggi + IP rendah"}
for k in (1, 2, 3):
    m = kelas == k
    if m.any():
        xs, zs = X[m], Z[m]
        print(f"{nama[k]:42s} x = {xs.min():5.1f}-{xs.max():5.1f} m, kedalaman {zs.min():4.1f}-{zs.max():4.1f} m, "
              f"rho median {np.median(rho[m]):6.1f} ohm.m, M median {np.median(M[m]):5.1f} mV/V")
print("Catatan: nilai dibuat untuk demonstrasi, bukan hasil survei lapangan.")

# ============================================================
# 3. VISUALISASI
# ============================================================
fig, ax = plt.subplots(3, 1, figsize=(10, 11))
im0 = ax[0].imshow(rho, extent=[0, 150, 40, 0], aspect="auto", cmap="jet", norm=LogNorm(vmin=15, vmax=700))
fig.colorbar(im0, ax=ax[0], label="Resistivitas (ohm.m)"); ax[0].set_title("Penampang resistivitas sintetis")
im1 = ax[1].imshow(M, extent=[0, 150, 40, 0], aspect="auto", cmap="jet", vmin=2, vmax=30)
fig.colorbar(im1, ax=ax[1], label="Chargeability (mV/V)"); ax[1].set_title("Penampang chargeability sintetis")
cmap = ListedColormap(["#e0e0e0", "#d62728", "#1f77b4", "#2ca02c"])
ax[2].imshow(kelas, extent=[0, 150, 40, 0], aspect="auto", cmap=cmap, vmin=0, vmax=3, interpolation="nearest")
ax[2].set_title("Klasifikasi zona (kombinasi resistivitas dan chargeability)")
ax[2].legend(handles=[Patch(color="#e0e0e0", label="Latar belakang"),
                      Patch(color="#d62728", label="IP tinggi + resistivitas rendah"),
                      Patch(color="#1f77b4", label="Resistivitas rendah + IP rendah"),
                      Patch(color="#2ca02c", label="Resistivitas tinggi + IP rendah")],
             loc="lower left", fontsize=8, ncol=2)
for a in ax: a.set_xlabel("Jarak lintasan (m)"); a.set_ylabel("Kedalaman (m)")
plt.tight_layout(); plt.show()
```

**Interpretasi hasil:** Metode Induced Polarization digunakan untuk mengenali zona yang dapat terpolarisasi, terutama mineral logam dan lempung, dengan membaca chargeability bersama resistivitas. Data contoh berasal dari konfigurasi Pole-Dipole sepanjang 150 m dengan kedalaman model 40 m. ## Eksekusi 5: Ground Penetrating Radar (GPR)

**Data sintetis:** radargram lintasan 15 m dengan jendela waktu 120 ns dan antena 400 MHz. Model berisi tiga reflektor berlapis (satu sedikit miring), dua objek terkubur yang menghasilkan pola hiperbola di sekitar jarak 4,5 m dan 10,5 m, atenuasi, dan noise. Data dibuat untuk demonstrasi, bukan data lapangan.

**Tujuan eksekusi:** memperlihatkan pemrosesan dasar radargram (dewow, penghapusan gelombang langsung, gain), estimasi kecepatan dari dua hiperbola difraksi, dan konversi waktu tempuh ke kedalaman.

```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# 1. PEMODELAN RADARGRAM SINTETIS (antena 400 MHz, lintasan 0-15 m, jendela 0-120 ns)
# ============================================================
rng = np.random.default_rng(21)
x = np.linspace(0, 15, 151)
t = np.linspace(0, 120, 301)                      # waktu tempuh dua arah (ns)
X, T = np.meshgrid(x, t)
v_true = 0.12                                     # m/ns (permitivitas relatif ~ 6), hanya untuk membangkitkan data
f0 = 0.40                                         # GHz

def ricker(tau):
    a = (np.pi*f0*tau)**2
    return (1 - 2*a)*np.exp(-a)

radar = 0.9*ricker(T - 5)                         # gelombang langsung (pita horizontal)
for z0, amp, dip in [(2.8, 0.50, 0.0), (3.8, 0.80, 0.04), (5.5, 0.45, 0.0)]:   # batas lapisan
    radar += amp*ricker(T - 2*(z0 + dip*x)/v_true)
objek = [(4.5, 1.1, 0.9), (10.5, 1.5, 1.0)]                 # (x, kedalaman, amplitudo): dua objek terkubur
for x_p, z_p, A in objek:
    t_hip = 2*np.sqrt(z_p**2 + (X - x_p)**2)/v_true
    radar += A*ricker(T - t_hip)
radar *= np.exp(-T/45)                            # atenuasi
radar += 0.015*rng.normal(size=radar.shape)       # noise

# ============================================================
# 2. PEMROSESAN SEDERHANA: dewow -> hapus gelombang langsung -> gain
# ============================================================
def dewow(a, w=25):
    k = np.ones(w)/w
    return a - np.apply_along_axis(lambda s: np.convolve(s, k, mode="same"), 0, a)

proc = dewow(radar)
proc = proc - np.where(T < 22, proc.mean(axis=1, keepdims=True), 0)   # hapus gelombang langsung (hanya zona awal agar reflektor datar tidak ikut hilang)
proc = proc*np.exp(T/70)                          # gain kompensasi atenuasi
proc[T < 11] = 0                                  # mute gelombang langsung

# ============================================================
# 3. ESTIMASI KECEPATAN DARI HIPERBOLA: t^2 = t0^2 + (2/v)^2 (x - xp)^2
# ============================================================
hasil = []
for (xc, tmin, tmax, hw) in ((4.5, 13, 40, 2.0), (10.5, 15, 41, 1.5)):
    cols = np.where(abs(x - xc) <= hw)[0]
    t_pick, x_pick = [], []
    for i in cols:
        w = (t > tmin) & (t < tmax)
        t_pick.append(t[w][np.argmax(proc[w, i])]); x_pick.append(x[i])
    t_pick, x_pick = np.array(t_pick), np.array(x_pick)
    x_apex = x_pick[np.argmin(t_pick)]
    slope, interc = np.polyfit((x_pick - x_apex)**2, t_pick**2, 1)
    v_est = 2/np.sqrt(slope); t0 = np.sqrt(interc)
    hasil.append((x_apex, t0, v_est, slope))
v_est = np.mean([h[2] for h in hasil])               # kecepatan rata-rata dari kedua hiperbola
eps_r = (0.2998/v_est)**2

print("GPR - DATA SINTETIS (antena 400 MHz, lintasan 15 m, jendela waktu 120 ns)")
for k, (xa, t0, v, sl) in enumerate(hasil, 1):
    print(f"Hiperbola {k}: puncak x ~ {xa:.1f} m, t0 ~ {t0:.1f} ns, v ~ {v:.3f} m/ns, kedalaman ~ {v*t0/2:.2f} m")
print(f"Kecepatan rata-rata         : {v_est:.3f} m/ns -> permitivitas relatif ~ {eps_r:.1f}")
for tt in (47, 70, 92):
    print(f"Reflektor datar t = {tt:3d} ns  : kedalaman ~ {v_est*tt/2:.2f} m")
print("Catatan: radargram merupakan simulasi untuk demonstrasi, bukan data lapangan.")

# ============================================================
# 4. VISUALISASI
# ============================================================
fig, ax = plt.subplots(1, 2, figsize=(14, 5.2), sharey=True)
ax[0].imshow(radar, extent=[0, 15, t.max(), 0], aspect="auto", cmap="gray", vmin=-0.5, vmax=0.5)
ax[0].set_title("Radargram mentah (sintetis)")
ax[1].imshow(proc, extent=[0, 15, t.max(), 0], aspect="auto", cmap="gray", vmin=-0.45, vmax=0.45)
for (xa, t0, v, sl), c, hw in zip(hasil, ("r", "c"), (2.0, 1.5)):
    xx = np.linspace(xa-hw, xa+hw, 80)
    ax[1].plot(xx, np.sqrt(t0**2 + sl*(xx-xa)**2), c+"--", lw=1.5, label=f"Fit hiperbola x = {xa:.1f} m (v = {v:.3f} m/ns)")
ax[1].set_title("Setelah dewow, penghapusan gelombang langsung, dan gain"); ax[1].legend(loc="lower right", fontsize=8)
for a in ax: a.set_xlabel("Jarak lintasan (m)")
ax[0].set_ylabel("Waktu tempuh dua arah (ns)")
sec = ax[1].secondary_yaxis("right", functions=(lambda tt: tt*v_est/2, lambda zz: zz*2/v_est))
sec.set_ylabel("Kedalaman (m)")
plt.tight_layout(); plt.show()
```

**Interpretasi hasil:** Metode GPR digunakan untuk memetakan lapisan dan objek dangkal dari pantulan gelombang elektromagnetik pada bidang kontras permitivitas. Data contoh berupa radargram sepanjang 15 m yang direkam dengan antena 400 MHz dan jendela waktu 120 ns. Radargram mentah didominasi oleh pita horizontal kuat di bagian atas yang berasal dari gelombang langsung, sedangkan refleksi yang lebih dalam tampak lemah karena atenuasi. Setelah dewow, penghapusan gelombang langsung, dan gain, refleksi di bagian dalam menjadi jelas. Gain juga menguatkan noise pada waktu besar sehingga bagian bawah radargram tampak lebih berbintik.

Dua pola hiperbola terlihat dengan puncak di sekitar jarak 4,4 m pada waktu sekitar 18,7 ns dan di sekitar jarak 10,5 m pada waktu sekitar 25 ns. Kelengkungan masing-masing hiperbola digunakan untuk mengestimasi kecepatan, dan hasilnya sekitar 0,122 dan 0,120 m/ns. Kedua nilai hampir sama, sehingga rata-ratanya, sekitar 0,121 m/ns, dapat dipakai untuk konversi kedalaman. Kecepatan ini setara dengan permitivitas relatif sekitar 6, yang sesuai dengan tanah atau pasir yang agak lembap. Dengan kecepatan tersebut, objek pertama berada pada kedalaman sekitar 1,1 m dan objek kedua sekitar 1,5 m. Pola hiperbola terbentuk oleh difraksi pada objek kecil yang terlokalisasi, seperti pipa, kabel, atau bongkah, tetapi jenis objeknya tidak dapat dipastikan hanya dari radargram. Jika objek berupa pipa yang memotong lintasan, orientasi lintasan terhadap pipa juga memengaruhi bentuk hiperbola.

Tiga reflektor yang relatif menerus terlihat pada waktu sekitar 47 ns, 63 sampai 73 ns, dan 92 ns. Dengan kecepatan rata-rata tadi, kedalamannya sekitar 2,9 m, sekitar 3,8 sampai 4,4 m, dan sekitar 5,6 m. Reflektor kedua memiliki amplitudo paling kuat dan miring ke arah kanan lintasan. Amplitudo besar menunjukkan kontras permitivitas yang besar, sehingga reflektor ini dapat berkaitan dengan muka air tanah atau batas dengan kadar air yang berbeda, tetapi interpretasi ini perlu dikonfirmasi dengan data sumur karena batas litologi yang tegas juga dapat menghasilkan refleksi kuat. Reflektor pertama dan ketiga lebih lemah dan dapat berkaitan dengan perubahan material atau tekstur tanah serta batas lapisan yang lebih dalam.

Kesimpulannya, radargram contoh menunjukkan struktur dangkal berlapis dengan tiga reflektor yang sedikit miring ke kanan, serta dua objek terkubur pada kedalaman sekitar 1,1 m dan 1,5 m. Keyakinan paling tinggi ada pada posisi lateral dan waktu tempuh objek, sedangkan keyakinan lebih rendah ada pada kedalaman absolut dan jenis objek karena bergantung pada kecepatan hasil fitting. Kecepatan tersebut diasumsikan seragam dan perlu dikalibrasi dengan penggalian terbatas atau data bor. Pada kondisi lapangan nyata, tanah lempung atau tanah yang sangat basah akan mengurangi kedalaman penetrasi secara nyata.
